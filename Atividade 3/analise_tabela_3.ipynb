# Analise da Tabela 3

Separacao da Tabela 3 em tres tabelas por regiao, homens e mulheres.

```python
import pandas as pd
```

## Leitura da Tabela 3

```python
arquivo = "Tabela_3_Areas_gerais_de_formacao_na_graduacao.xlsx"

# A planilha tem cabecalho em varias linhas, por isso lemos sem definir cabecalho.
tabela3 = pd.read_excel(arquivo, header=None)
tabela3.head(8)
```

## Separacao das regioes

```python
regioes = ["Norte", "Nordeste", "Sudeste", "Sul", "Centro-Oeste"]

dados_regioes = tabela3[tabela3[0].isin(regioes)].copy()
dados_regioes
```

## Funcao para criar e salvar cada tabela

```python
def criar_tabela(coluna_homens, coluna_mulheres, nome_arquivo):
    tabela = dados_regioes[[0, coluna_homens, coluna_mulheres]].copy()
    tabela.columns = ["Regiao", "Homens", "Mulheres"]
    tabela["Homens"] = pd.to_numeric(tabela["Homens"], errors="coerce")
    tabela["Mulheres"] = pd.to_numeric(tabela["Mulheres"], errors="coerce")

    estatisticas = tabela[["Homens", "Mulheres"]].describe()

    with pd.ExcelWriter(nome_arquivo, engine="openpyxl") as writer:
        tabela.to_excel(writer, sheet_name="Dados", index=False)
        estatisticas.to_excel(writer, sheet_name="Estatisticas_describe")

    return tabela, estatisticas
```

## 1. Homens e mulheres por regiao com curso superior

```python
tabela_curso_superior, estatisticas_curso_superior = criar_tabela(
    coluna_homens=2,
    coluna_mulheres=3,
    nome_arquivo="tabela_curso_superior.xlsx"
)

tabela_curso_superior
```

```python
estatisticas_curso_superior
```

## 2. Homens e mulheres por regiao com curso em Ciencia, Tecnologia, Engenharias e Matematica

```python
tabela_ctem, estatisticas_ctem = criar_tabela(
    coluna_homens=5,
    coluna_mulheres=6,
    nome_arquivo="tabela_ctem.xlsx"
)

tabela_ctem
```

```python
estatisticas_ctem
```
