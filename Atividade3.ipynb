```python
from pathlib import Path
import pandas as pd
from IPython.display import display

ARQUIVO_ENTRADA = Path("Tabela_3_Areas_gerais_de_formacao_na_graduacao.xlsx")
REGIOES = ["Norte", "Nordeste", "Sudeste", "Sul", "Centro-Oeste"]

dados_brutos = pd.read_excel(ARQUIVO_ENTRADA, sheet_name="Sheet1", header=None)
print(f"Planilha carregada: {dados_brutos.shape[0]} linhas e {dados_brutos.shape[1]} colunas.")
```

```python
linhas_regioes = dados_brutos[0].isin(REGIOES)

def criar_tabela(coluna_homens, coluna_mulheres):
    tabela = dados_brutos.loc[linhas_regioes, [0, coluna_homens, coluna_mulheres]].copy()
    tabela.columns = ["Região", "Homens", "Mulheres"]
    tabela[["Homens", "Mulheres"]] = tabela[["Homens", "Mulheres"]].apply(pd.to_numeric)
    tabela["Região"] = pd.Categorical(tabela["Região"], categories=REGIOES, ordered=True)
    return tabela.sort_values("Região").reset_index(drop=True)

tabela_curso_superior = criar_tabela(2, 3)
tabela_ctem = criar_tabela(5, 6)
tabela_educacao_saude = criar_tabela(8, 9)
```

```python
def salvar_excel(nome_arquivo, tabela, estatisticas):
    with pd.ExcelWriter(nome_arquivo, engine="openpyxl") as writer:
        tabela.to_excel(writer, sheet_name="Dados", index=False)
        estatisticas.to_excel(writer, sheet_name="Estatisticas_describe")

arquivos_saida = {
    "tabela_curso_superior.xlsx": (tabela_curso_superior, estatisticas_curso_superior),
    "tabela_ctem.xlsx": (tabela_ctem, estatisticas_ctem),
    "tabela_educacao_servicos_saude.xlsx": (tabela_educacao_saude, estatisticas_educacao_saude),
}

for nome, (tabela, estatisticas) in arquivos_saida.items():
    salvar_excel(nome, tabela, estatisticas)
    print(f"Arquivo salvo: {Path(nome).resolve()}")
```
