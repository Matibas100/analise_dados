# População brasileira: 2010 e 2022

Agregações por estado e município, ordenadas pela variação absoluta da população entre 2010 e 2022.

```python
import pandas as pd

FONTE_DADOS = 'CD2022_Populacao_2010_Compatibilizada_20231222.xlsx'

dados_brutos = pd.read_excel(FONTE_DADOS, sheet_name=0, skiprows=2)
dados_brutos = dados_brutos[dados_brutos['COD. UF'].notna()].copy()
dados_brutos.columns = ['OMITIR', 'ESTADO', 'ID_ESTADO', 'ID_MUNICIPIO', 'NOME_MUNICIPIO', 'HABITANTES_2010_INICIAL', 'HABITANTES_2010_AJUSTADO', 'HABITANTES_2022']

censo_anterior = 'HABITANTES_2010_AJUSTADO'
censo_atual = 'HABITANTES_2022'
dados_brutos['Variacao_Absoluta'] = dados_brutos[censo_atual] - dados_brutos[censo_anterior]

dados_brutos.head()
```

```python
consolidado_estado = (
    dados_brutos.groupby('ESTADO')[[censo_anterior, censo_atual, 'Variacao_Absoluta']]
      .sum()
      .reset_index()
      .sort_values('Variacao_Absoluta', ascending=False)
)

consolidado_estado.to_csv('populacao_por_estado.csv', sep=';', index=False, encoding='utf-8-sig')
consolidado_estado
```

```python
ranking_municipios = (
    dados_brutos[['ESTADO', 'ID_ESTADO', 'ID_MUNICIPIO', 'NOME_MUNICIPIO', censo_anterior, censo_atual, 'Variacao_Absoluta']]
      .sort_values('Variacao_Absoluta', ascending=False)
)

ranking_municipios.to_csv('populacao_por_municipio.csv', sep=';', index=False, encoding='utf-8-sig')
ranking_municipios.head()
```
